```python
#all data from csv files except the rows for the a.out and software columns

import pandas as pd

# list of file paths
df_list = []
mergedDataframe = pd.DataFrame()

for year in ["2026", "2025", "2024", "2023", "2022", "2021"]:
    for month in range(1, 13):
        # :02d turns the number 5 into "05" to match formatting
        url = f"https://raw.githubusercontent.com/ARCHER2-HPC/usage-data/main/allusers/{year}/{month:02d}/{year}-{month:02d}_scua_CodeSize_weighted.csv"

        if (year == "2023" and month < 3) or (year == "2022" and month > 9):
            url = f"https://raw.githubusercontent.com/ARCHER2-HPC/usage-data/main/allusers/{year}/{month:02d}/{year}-{month:02d}_CodeSize_weighted.csv"
        elif (year == "2022" and month <= 9) or (year == "2021" and month == 12):
            url = f"https://raw.githubusercontent.com/ARCHER2-HPC/usage-data/main/allusers/{year}/{month:02d}/{year}-{month:02d}_stats_by_usage.csv"

        try:
            df = pd.read_csv(url)

            df["Year"] = pd.Series([year] * len(df), dtype="category")
            df["Month"] = pd.Series([month] * len(df), dtype="category")

            #creating a column for season and populating it
            if (month == 12) or (month >=1 and month <=2):
                df["Season"] = pd.Series(["Winter"] * len(df), dtype="category")
            elif (month >=3 and month <=5):
                df["Season"] = pd.Series(["Spring"] * len(df), dtype="category")
            elif (month >=6 and month <=8):
                df["Season"] = pd.Series(["Summer"] * len(df), dtype="category") 
            else:
                df["Season"] = pd.Series(["Autumn"] * len(df), dtype="category")
            
            df["Nodeh"] = df["Nodeh"].fillna(0)
            df.loc[df["Software"].isin(["Unidentified", "Unknown"]),"Software"] = "Unknown"

            
            #dropping off all data for rows for the overall and a.out software
            drop = "Overall"
            DROP2 = "a.out"
            DROP3 = "No srun"
            df = df[(df["Software"] != drop) & (df["Software"] != DROP2) & (df["Software"] != DROP3)]

            df_list.append(df)#appending csv files to array

        except Exception as e:
            print(f"An error occurred while reading {url}: {e}") #error message

# updated dataframe and outputs

mergedDataframe = pd.concat(df_list, axis=0, ignore_index=True)
# season_order = ["Winter", "Spring", "Summer", "Autumn"]

# mergedDataframe["Season"] = pd.Categorical(
#     mergedDataframe["Season"],
#     categories=season_order,
#     ordered=True
# )

print(mergedDataframe)
mergedDataframe.to_csv("archer2_HPC_data_without_overall_and_aOut.csv")
```
