```python
#yearly trend for nodeh vs software for the top 5 softwares across 2021-2026 for eqach month in each yuear

"""
finding the 10 software with most nodeh from 2021-2026
find the aggregarted values for noideh for each software for each of these years
plot nodeh vs software as a line graph with software as lines months as x axis and nodeh as y axis for each year
"""


import pandas as pd
import matplotlib.pyplot as plt

# dataframe
hpcData = pd.read_csv("archer2_HPC_data_without_overall_and_aOut.csv")

# top 5 software overall (2021–2026)
overall_top10 = (
    hpcData.groupby("Software")["Nodeh"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top10_names = overall_top10.index.tolist()
print("Top 5 software overall:")
print(overall_top10)

# filter dataset to only those top 10 software
filtered = hpcData[hpcData["Software"].isin(top10_names)]

# loop through each year and create a monthly trend plot
for yearName, yearGroup in filtered.groupby("Year"):

    # sum up monthly Nodeh for each software
    monthly_trend = (
        yearGroup.groupby(["Month", "Software"])["Nodeh"]
        .mean()
        .reset_index()
    )

    # months = rows, software = columns
    pivot_df = monthly_trend.pivot_table(
        index="Month",
        columns="Software",
        values="Nodeh",
        aggfunc="mean"
    ).fillna(0)

    # graph for each year
    pivot_df.plot(kind="line", figsize=(12, 9), linewidth=1.2, marker="*", cmap="tab10", markersize=7)

    plt.title(f"Monthly Nodeh Trend for Top 5 Software — {yearName}")
    plt.xlabel("Months")
    plt.ylabel("Node * Hour")
    plt.grid(axis="y", linestyle="-", alpha=0.6)
    plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

hpcData = pd.read_csv("archer2_HPC_data_without_overall_and_aOut.csv")

# Top 5 software overall
overall_top5 = (
    hpcData.groupby("Software")["Nodeh"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

top5_names = overall_top5.index.tolist()

# Filter to top 5
filtered = hpcData[hpcData["Software"].isin(top5_names)]

# Create Year-Month combined axis
filtered["YearMonth"] = (
    filtered["Year"].astype(str) + "-" + filtered["Month"].astype(str)
)

# Compute monthly Nodeh trend for each software
trend = (
    filtered.groupby(["YearMonth", "Software"])["Nodeh"]
    .sum()
    .reset_index()
)

# Pivot: rows = YearMonth, columns = Software
pivot_df = trend.pivot_table(
    index="YearMonth",
    columns="Software",
    values="Nodeh",
    aggfunc="sum"
).fillna(0)

# Plot monthly trend across all years
pivot_df.plot(
    kind="line",
    figsize=(14, 8),
    marker="*",
    linewidth=1.5,
    cmap="tab10"
)

plt.xticks(rotation=90)
plt.xlabel("Year-Month")
plt.ylabel("Nodeh")
plt.title("Monthly Nodeh Trend for Top 5 Software Across All Years (2021–2026)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
